### 2D Arrays Review Guide

## 📌 What is a 2D Array?
A 2D array is an array of arrays, often visualized as a **grid** or **matrix** of rows and columns.

### Syntax
```java
// Declaration
int[][] matrix;

// Initialization
matrix = new int[3][4]; // 3 rows, 4 columns

// Declaration + Initialization
int[][] matrix = {
    {1, 2, 3},
    {4, 5, 6},
    {7, 8, 9}
};
