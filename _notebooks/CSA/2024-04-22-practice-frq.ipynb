# Overview of AP CSA FRQ Types

## 1. Methods and Control Structures
This question tests your ability to write methods using basic control structures like conditionals (`if`/`else`) and loops (`for`, `while`). You'll be asked to implement a method that solves a specific problem by applying logic to manipulate variables or control the flow of the program. The problem may involve tasks like counting, filtering, or modifying data. It's important to be familiar with common programming patterns, such as breaking down problems into smaller steps, as well as handling edge cases and ensuring correctness with loops and conditionals.

### Key Concepts:
- Conditional statements (`if`/`else`, `switch`)
- Looping structures (`for`, `while`, `do-while`)
- Data manipulation through logic
- Algorithm design

## 2. Classes
In this question, you'll be required to write or modify a class, focusing on key object-oriented programming (OOP) concepts like fields, constructors, methods, and encapsulation. You may need to create new methods to perform certain actions or modify existing ones to meet specific requirements. The problem may also involve understanding the relationship between objects and their attributes. Be prepared to implement methods that interact with fields, validate input, and manage the state of an object. Understanding how to structure and design a class effectively is crucial.

### Key Concepts:
- Class design and structure
- Constructors and field initialization
- Method creation and modification
- Object-oriented principles (encapsulation, abstraction)
- Interacting with object attributes and state

public class Book {
    private String title;
    private String author;
    private int pages;

    public Book(String title, String author, int pages) {
        this.title = title;
        this.author = author;
        this.pages = pages;
    }

    public String getDescription() {
        return title + " by " + author + ", Pages: " + pages;
    }
}

## 3. Array/ArrayList
This FRQ will assess your understanding of arrays and `ArrayLists`—two of the most common data structures in Java. You'll be asked to manipulate arrays or `ArrayLists` in various ways, such as searching for elements, summing values, or applying specific transformations. The problem could involve tasks like finding the largest or smallest value in an array, filtering out certain elements, or performing operations that require traversal and modification of the array. You'll need to handle edge cases (e.g., empty arrays) and consider the efficiency of your solution.

### Key Concepts:
- Array/ArrayList manipulation
- Traversing and modifying elements
- Searching and sorting algorithms
- Handling edge cases and null values
- Indexing and bounds checking

public int[] findMinMax(int[] arr) {
    int min = arr[0];
    int max = arr[0];
    
    for (int i = 1; i < arr.length; i++) {
        if (arr[i] < min) {
            min = arr[i];
        }
        if (arr[i] > max) {
            max = arr[i];
        }
    }
    
    return new int[]{min, max};
}




## 4. 2D Array
In a 2D array question, you'll work with an array of arrays (e.g., `int[][]` or `String[][]`). These problems typically require nested loops to access and manipulate elements in the 2D structure. You might be asked to implement algorithms that traverse rows and columns, search for patterns, or calculate sums. 2D arrays are often used to represent matrices or grids, and problems may involve pathfinding, pattern recognition, or applying mathematical operations like matrix multiplication.

### Key Concepts:
- Working with 2D arrays (`int[][]`, `String[][]`, etc.)
- Nested loops and indexing
- Matrix manipulation or grid traversal
- Searching or calculating values in a 2D structure
- Handling multidimensional data

## 5. Inheritance (Occasionally Tested)
Inheritance questions test your understanding of class hierarchies and polymorphism. You may be asked to design or modify classes that inherit from other classes, with the goal of reusing or extending functionality. This could involve overriding methods, using `super` to call a parent class’s constructor, or designing a class that leverages the benefits of inheritance (e.g., code reuse). While not as common as other question types, it’s important to understand the core principles of inheritance in Java and how it can be used to create more flexible and maintainable code.

### Key Concepts:
- Class inheritance and extending classes
- Method overriding and polymorphism
- Constructor chaining with `super()`
- Designing flexible and reusable code
- Relationships between parent and child classes

## 6. Recursion (Occasionally Tested)
Recursion questions evaluate your ability to write methods that call themselves to solve problems. These problems often involve tasks like searching or calculating values by breaking the problem into smaller, more manageable pieces. You’ll need to define a base case to stop the recursion and a recursive case to continue the process. Common examples include calculating factorials, summing numbers in a list, or solving problems that require a divide-and-conquer approach. Understanding the stack frame, base cases, and recursive logic is essential for solving these problems.

### Key Concepts:
- Recursive problem-solving
- Base case and recursive case definition
- Divide-and-conquer approach
- Stack frame management
- Debugging recursion and identifying infinite loops
