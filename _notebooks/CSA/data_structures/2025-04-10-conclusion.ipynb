### Challenge Question (Bonus +0.01)

**Question:**  
You need to store a collection of student IDs where:
- Order **doesn't matter**
- You must prevent **duplicates**
- You often need to **check if an ID exists**

Which collection type would be most efficient to use and why?


```java
import java.util.ArrayList;
import java.util.List;

public class ListExample {
    public static List<Integer> getEvenNumbers(List<Integer> numbers) {
        List<Integer> evens = new ArrayList<>();
        for (int i = 0; i < numbers.size(); i++) {
            int num = numbers.get(i);
            if (num % 2 == 0) {
                evens.add(num);
            }
        }
        return evens;
    }
}
```

```java
import java.util.HashSet;
import java.util.Set;

public class SetExample {
    public static Set<String> getIntersection(Set<String> set1, Set<String> set2) {
        Set<String> result = new HashSet<>();
        for (String s : set1) {
            if (set2.contains(s)) {
                result.add(s);
            }
        }
        return result;
    }
}
```

```java
import java.util.ArrayDeque;
import java.util.Deque;

public class DequeExample {
    public static void simulateLine() {
        Deque<String> line = new ArrayDeque<>();

        // 1. Add 3 customers to the end
        line.addLast("Customer A");
        line.addLast("Customer B");
        line.addLast("Customer C");

        // 2. Add 1 VIP customer to the front
        line.addFirst("VIP Customer");

        // 3. Remove the customer at the front
        line.removeFirst();

        // 4. Show the front and back of the line
        System.out.println("Front: " + line.peekFirst());
        System.out.println("Back: " + line.peekLast());

        // 5. Print the size of the line
        System.out.println("Size: " + line.size());
    }
}
```

The best collection type to use is a **HashSet**.

**Reason:**
- **Order doesn't matter** → HashSet does not maintain any specific order.
- **No duplicates allowed** → HashSet automatically prevents duplicate entries.
- **Fast existence checks** → HashSet provides constant-time performance (O(1) on average) for `contains()` operations.

Therefore, `HashSet` is the most efficient choice for storing unique student IDs when order is irrelevant and quick lookups are needed.
